```python
from collections import defaultdict

class Graph:
    def __init__(self):
        self.graph = defaultdict(list)

    def addEdge(self, u, v):
        self.graph[u].append(v)

    def DFSUtil(self, v, visited):
        visited[v] = True
        print(v, end = ' ')
        for i in self.graph[v]:
            if not visited[i]:
                self.DFSUtil(i, visited)

    def DFS(self, v):
        # Initialize visited array based on the maximum vertex index present in the graph
        # or the number of unique vertices, to avoid IndexError.
        # A safer approach for visited array initialization.
        max_vertex = max(max(self.graph.keys()) if self.graph else 0, max(item for sublist in self.graph.values() for item in sublist) if self.graph else 0)
        visited = [False] * (max_vertex + 1)
        self.DFSUtil(v, visited)

# Create a graph instance
g = Graph()

# Add edges
g.addEdge(0, 1)
g.addEdge(0, 2)
g.addEdge(1, 2)
g.addEdge(2, 0)
g.addEdge(2, 3)
g.addEdge(3, 3)

# Perform DFS starting from vertex 2
print("Following is DFS from (starting from vertex 2)")
g.DFS(2)
```

```python
from collections import defaultdict

class Graph:
    def __init__(self):
        self.graph = defaultdict(list)

    def addEdge(self, u, v):
        self.graph[u].append(v)

    def BFS(self, s):
        # To handle cases where 's' might be greater than any existing vertex,
        # or if there are disconnected components. Adjust visited array size.
        max_vertex = max(max(self.graph.keys()) if self.graph else 0, max(item for sublist in self.graph.values() for item in sublist) if self.graph else 0)
        # Ensure visited array can cover the starting vertex 's' if it's larger
        visited_size = max(max_vertex + 1, s + 1)
        visited = [False] * visited_size

        queue = []
        queue.append(s)
        visited[s] = True

        while queue:
            s = queue.pop(0)
            print(s, end=" ")

            for i in self.graph[s]:
                if not visited[i]:
                    queue.append(i)
                    visited[i] = True

# Create a graph instance
g = Graph()

# Add edges
g.addEdge(0, 1)
g.addEdge(0, 2)
g.addEdge(1, 2)
g.addEdge(2, 0)
g.addEdge(2, 3)
g.addEdge(3, 3)

print("Following is Breadth First Traversal" " (starting from vertex 2)")
g.BFS(2)
```
